```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from collections import Counter

# Load the uploaded CSV file into a DataFrame called df
df = pd.read_csv('netflix_titles.csv')

# Display the first 5 rows
display(df.head())
```

### Chart 1: Total Count of Movies vs TV Shows

```python
plt.figure(figsize=(8, 6))
sns.countplot(data=df, x='type', palette='Set2', hue='type', legend=False)
plt.title('Total Count of Movies vs TV Shows', fontsize=14, fontweight='bold')
plt.xlabel('Type', fontsize=12)
plt.ylabel('Count', fontsize=12)
plt.tight_layout()
plt.show()
```

### Chart 2: Percentage of Movies vs TV Shows

```python
plt.figure(figsize=(8, 6))
type_counts = df['type'].value_counts()
plt.pie(type_counts, labels=type_counts.index, autopct='%1.1f%%', startangle=90, colors=sns.color_palette('Set2', len(type_counts)))
plt.title('Percentage of Movies vs TV Shows', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```
